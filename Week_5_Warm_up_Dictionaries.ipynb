In this tutorial you'll learn about python's dictionaries. You'll be able to access the information stored in a dictionary and modify information based on your needs. With dictionaries you'll be able to model a variety of objects. Additionally, you'll learn how to loop through the data in the dictionary, nest dictionaries inside lists, lists inside dictionaries and dictionaries inside other dictionaries. In python, a dictionary is a collection of key-value pairs. Each key is associated with a value and you can access the value associated with that key. Any object that python recognizes can be used as a value in a dictionary. A dictionary is wrapped in braces (`{}`) with a series of key-value pairs inside the braces.

```python
#Build a simple dictionary

plant_0 = {}

plant_0['name']= 'monstera deliciosa'
plant_0['color'] = 'yellow'
plant_0['watering'] = 1

print(plant_0)
```

## Changing values in a dictionary

```python
plant_0 = {'color': 'green'}
print(f"The plant is {plant_0['color']}.")

plant_0['color'] = 'yellow'
print(f"The plant is now {plant_0['color']}.")
```

A more complex example of a dictionary could be tracking the growth of a plant. This could be applied in a game setting. This also works well if you need to change values in the dictionary. Just change the criteria based on your needs

```python
plant_0 = {'x_position': 0, 'y_position': 2, 'height': 'medium'}
print(f"Original position: {plant_0['x_position']}")

# Grow the plant.
# Determine how high the plant grew based on its height.
if plant_0['height'] == 'small':
    x_increment = 0.5
elif plant_0['height'] == 'medium':
    x_increment = 1
else:
    # This must be a fast - growing plant.
    x_increment = 1.5

# The new position is the old position plus the increment.
plant_0['x_position'] = plant_0['x_position'] + x_increment
print(f"New position: {plant_0['x_position']}")
```

## Removing key-value pairs from a dictionary
Sometimes you'll want to remove key-value pairs from a dictionary. You can do this using the `del` statement. Be mindful that if you delete a key-value pair it will be gone forever.

```python
plant_0 = {'color': 'green', 'watering': 1}
print(plant_0)

del plant_0['watering']
print(plant_0)
```

### Creating a dictionary of similar objects
Instead of creating a dictionary based on one object, i.e. plant example. You could also create dictionaries to store one kind of information based on multiple objects. When you know you’ll need more than one line to define a dictionary, press ENTER after the opening brace. Then indent the next line one level (four spaces) and write the first key-value pair, followed by a comma. From this point forward when you press ENTER, your text editor should automatically indent all subsequent key-value pairs to match the first key-value pair.

```python
#We can look up Sarah's favorite ML library like this:
favorite_ml_libraries = {
    'jen': 'pandas',
    'sarah': 'sklearn',
    'adam': 'seaborn',
    'phil': 'matplotlib',
    }

library = favorite_ml_libraries['sarah'].title()
print(f"Sarah's favorite ml library is {library}.")
```

For dictionaries specifically, you can use the `get()` method to set a default value that will be returned if the requested key doesn’t exist.This will prevent an error from occuring since you are specifing to python what you would like to see if the criteria are not met. If you think the  the key you’re asking for might not exist, consider using the `get()` method instead of the square bracket notation.

```python
plant_0 = {'name': 'monstera deliciosa', 'color': 'green'}

watering_value = plant_0.get('watering', 'No watering value assigned.')
print(watering_value)
```

### Looping through Dictionaries
Since dictionaries can contain a lot information; there are many ways to loop through them.This code tells Python to loop through each key-value pair in the dictionary. As it works through each pair the key is assigned to the variable: name, and the value is assigned to the variable: library. Here is an example of how you can loop through a dictionary:

```python
favorite_ml_libraries = {
    'jen': 'pandas',
    'sarah': 'sklearn',
    'adam': 'seaborn',
    'phil': 'matplotlib',
    }
for name, library in favorite_ml_libraries.items():
    print(f"{name.title()}'s favorite ml library is {library.title()}.")
```

```python
#we can only print the keys
for name in favorite_ml_libraries.keys():
    print(name.title())
```

```python
#the output is the same. It's the default
for name in favorite_ml_libraries:
    print(name.title())
```

```python
#we can only print the values
friends = ['phil', 'sarah']
for name in favorite_ml_libraries:
    print(f"Hi {name.title()}.")

    if name in friends:
        library = favorite_ml_libraries[name].title()
        print(f"\t{name.title()}, I see you share my interest in {library}!")
```

### Looping through values in a dictionary with `set()`. 
To see each value in a dictionary without repeating it, use the `set()` function to convert the dictionary into a set. Then loop through the set. A set is an unordered collection of unique items.


```python
# getting the values from a dictionary and checking for repeats
for library in set(favorite_ml_libraries.values()):
    print(library.title())
```

### Working with Nesting multiple dictionaries or storing a list of items in a dictionary
Nesting is referring to storing information in a dictionary that contains other dictionaries or lists. For example, you can store a list of dictionaries in a dictionary.

### A List of Dictionaries
Lists of dictionaries are useful when you would like to store information on multiple objects instead of just one or two. For example, you could store a list of plants in which each plant is a dictionary of each plant. We can generate a collection of plants with `range()`The plants will have the same set of attributes, but these could be changed.

```python
#make an empty list

plants = []
#store 20 plants
for plant in range(20):
    new_plant={'name': 'aloe', 'color': 'green', 'watering': 'low'}
    plants.append(new_plant)
#show the first 5 plants
for plant in plants [:5]:
    print(plant)
print('...')

#Show the 5 plants
print (f"Total number of plants: {len(plants)}")
```

```python
#change the first 5 plants to be yellow, medum watering
for plant in plants[:5]:
    if plant['color'] == 'green':
        plant['color'] = 'yellow'
        plant['watering'] = 'medium'
        
        
#show the first 5 plants
for plant in plants[:5]:
    print(plant)
print('...')
```

### A List in a Dictionary
Sometimes it's useful to put a list inside a dictionary. For example, you can store multiple aspects of an object instead of just one.

```python
#store information about cookie preparation

cookie_recipe = {
    'type': 'chocolate chip',
    'ingredients': ['butter', 'chocolate chips', 'flour', 'eggs'],}

#Restate the recipe
print(f"Here's how to make {cookie_recipe['type']} cookies:")
for ingredient in cookie_recipe['ingredients']:
    print(f"- {ingredient}")
```

### A Dictionary in a Dictionary
This is another example of nesting dictionaries. We can store information on multiple objects in a dictionary, i.e. users and their user names. We can access this information by looping through the usernames and the dictionary of information associated with each username.

```python
users = {
    'aeinstein': {
        'first': 'albert',
        'last': 'einstein',
        'location': 'princeton',
        },

    'mcurie': {
        'first': 'marie',
        'last': 'curie',
        'location': 'paris',
        },

    }

for username, user_info in users.items():
    print(f"\nUsername: {username}")
    full_name = f"{user_info['first']} {user_info['last']}"
    location = user_info['location']

    print(f"\tFull name: {full_name.title()}")
    print(f"\tLocation: {location.title()}")
```
