## Conditional Tests
Conditional tests are the basis for if statements. Python uses the values `True` and `False` to decide whether the code in an if statement should be executed. If a conditional test evaluates to `True`, Python executes the code that follows the if statement. If it evaluates to `False`, Python ignores the code following the if statement.

```python
team= 'mercedes' #set the value of team to mercedes
team=='mercedes' # Is the name of the team mercedes?
```

Testing for equality like the above example is case sensitive. This is useful when you want to  differentiate text with capitalization. However, if you would like to detect both lower case and upper case variations, you would need to convert an upper case word to lower case. This doesn't change the value of the orignial variable.

```python
team= 'Red bull'
team.lower()=='red bull'
team
```

## Logical Operators
Sometimes you may need to check multiple conditions at once. As you combine two or more conditional tests, they would all need to meet the requirements for the statement to evaluate to `true` . If either test or both tests fail the expression will evaluate to `False`. The and operator checks if all conditions are `True`. The or operator checks if at least one condition is `True`. 

```python
#using and
age = 24
citizen = True

if age >= 18 and citizen:
    print("You can vote!")
```

```python
#using or 
age = 24
citizen = False

if age <= 18 or citizen== False:
    print("You can't vote yet!")
```

### If Statements. 
The simplest type of `if` statement has one test and one action:
if conditional_test:
    do something
You can put any conditional test in the first line of an `if` statement. If the test evaluates to `True`, the code following the `if` statement is executed. If the test evaluates to `False`, the code following the `if` statement is skipped.The following will test whether
you are over 20 years old. The indented lines after the if statement will execute if the test passes, and the whole block of code will be ignored if the conditional test doesn't pass. 

```python
age = 17
if age >= 17:
    print("You can drive a car!")
    print("Have you received a driver's license yet?")
```

### If-else Statements
If you want to take one action when a conditional test passes and another action when a conditional test fails, you can use an `if` statement followed by an `else` statement.The `else` statement will allow you to specify an action or a set of actions to be executed if the conditional test evaluates to `False`.For example:


```python
age = 16
if age >= 17:
    print("You can drive a car!")
    print("Have you received a driver's license yet?")
else:
    print("You can't drive yet!")
    
```

### If-elif-else Statements
If you would like to test more than two possible outcomes, you can use an `if`- `elif`- `else` statement. Only one block of code will be executed, based on the outcome of the conditional test. Consider different weather conditions and what you might do in certain situations. Also, you can add as many elif blocks as you like depending on the number of conditions you desire to include. For example:

```python
weather = "sunny"
if weather == "sunny":
    print ("Go out and play!")
elif weather == "cloudy":
    print("Stay inside")
else:
    print("Go somewhere else")
```

The `else` block is a catch-all statement. It matches any condition that wasn't matched by an `if` or `elif` block. If you have a specific final condition to test for, consider using a final `elif` block and omit the `else` block. This will ensure that your code will try out all of the possible conditions. Sometimes you may need to check all conditions of interest. Then you would use a series of simple `if` statements. This is helpful when more than one condition is true and you would like to act on it. For example:

```python
recipe_ingredients = ["honey", "strawberries", "vanilla"]
if "strawberries" in recipe_ingredients:
    print("adding strawberries")
if "chocolate" in recipe_ingredients:
    print("I'm going to make chocolate cake")
if "vanilla" in recipe_ingredients:
    print("adding vanilla")
print ("I'm going to make strawberry shortcake.")
    
```

### Using If statements with Lists
You can specify special values to be treated differently than other values in the list. We'll continue with the recipe example in the next section.By inserting the "ingredient" in the `for` loop, we can print out each item in the list.

```python
recipe_ingredients = ["honey", "strawberries", "vanilla"]

for ingredient in recipe_ingredients:
        print(f"Add {ingredient}.")
print ("I'm going to make strawberry shortcake.")
```

It's possible we may run out of an ingridient. To handle this situation, we can use an `else` block. The `else` block will be executed if the `for` loop completes without encountering a break.

```python
recipe_ingredients = ["honey", "strawberries", "vanilla"]

for ingredient in recipe_ingredients:
        if ingredient == "honey":
            print("I don't have honey. Skipping this ingredient.")
        else:
            print(f"Add {ingredient}.")
print ("I'm going to make strawberry shortcake.")
```

We can add more compexity when we have multiple lists. What if you change your mind and want to make cookies? We can store a list of available ingredients you have on hand in the first list. The second list will store the ingredients needed to make cookies.

```python
ingredients_at_home = ["honey", "strawberries", "vanilla", "flour", "eggs", "butter"]

recipe_ingredients = ["flour", "eggs", "butter"]

for ingredient in recipe_ingredients:
        if ingredient in ingredients_at_home:
            print(f"I have {ingredient}.")
        else:
            print(f"Oops, I don't have {ingredient}.")
print ("I'm going to make cookies.")
```
